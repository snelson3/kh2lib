```python
# First create the kh2lib object
from kh2lib.kh2lib import kh2lib
import os, json
if not os.path.isdir("workspace"):
    os.mkdir("workspace")

lib = kh2lib(game="bbs")
lib.gamedir = 'C:\\Users\\12sam\Desktop\\KH_Games\\bbs'
    
bbs_in_dir = lib.gamedir
bbs_out_dir = os.path.join(lib.gamedir, "..", "bbs_arcs")
```

```python
# Extract all bars
# problem - output path shouldn't include the .
for root, dirs, files in os.walk(bbs_in_dir):
    path = root.split(os.sep)
    for file in files:
        fn = os.path.join(root, file)
        ext = open(fn, "rb").read(3)
        if ext == b'OLO':
            print("olo at",fn)
        if ext == b'ARC':
            subfn = fn.replace(bbs_in_dir, '')[1:]
            # make outfn, ensure output folder exists
            if len([c for c in subfn if c == os.sep]) == 0:
                subfn = os.path.join("root", *subfn.split(os.sep))
            outfn = os.path.join(bbs_out_dir, '.'.join(subfn.split(".")[:-1]))
            if not os.path.isdir(os.path.dirname(outfn)):
                os.makedirs(os.path.dirname(outfn))
            lib.editengine.arc_extract(fn, outfn)
```
