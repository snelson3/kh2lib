```python
# First create the kh2lib object
from kh2lib.kh2lib import kh2lib
import os, json
lib = kh2lib()
if not os.path.isdir("workspace"):
    os.mkdir("workspace")

game = "KH2"
basepth = os.path.join(os.environ["KHGAMES_PATH"], game)
kh2_in_dir = os.path.join(basepth, game)
kh2_out_dir = os.path.join(basepth, "bars")
```

```python
# Extract bdx_files
import shutil
subfiles = {
    "bdx_built": lib.openkh.bdx_extract
}
c = 0
failures = []
for sf in subfiles:
    c = 0
    failures = []
    subfile_out_dir = os.path.join(basepth, "subfiles", sf)
    for root, dirs, files in os.walk(subfile_out_dir):
        path = root.split(os.sep)
        for f in files:
            fn = os.path.join(root, f)
            if fn.endswith(".bdx".format(sf)):
                c += 1
                outfn = fn.replace('bdx_built', 'bdscript_decompiled')
                outfn = outfn.replace('.bdx', '.bdscript')
                if len([c for c in outfn if c == os.sep]) == 1:
                    outfn = os.path.join("root", *subfn.split(os.sep))
                if not os.path.isdir(os.path.dirname(outfn)):
                    os.makedirs(os.path.dirname(outfn))
                subfiles[sf](fn, outfn)
                if not os.path.exists(outfn):
                    failures.append(outfn)
print("{}/{}".format(len(failures), c))
```

```python
# Recompile bdx_files
# loop all files
# try to build
# warning if build fails
# check to see if byte for byte the same, just count
import shutil,random
bdscript_dir = os.path.join(basepth, "subfiles", "bdx")
outdir = os.path.join(basepth, "tmpbuild")
if os.path.isdir(outdir):
    shutil.rmtree(outdir)

c = 0
broken_builds = []
byte_identical = [0,0]
for root, dirs, files in os.walk(subfile_out_dir):
    path = root.split(os.sep)
    for f in files:
        if not f.endswith(".bdscript"):
            continue
        c += 1
        fn = os.path.join(root, f)
        input_fn = fn.replace("bdx_built", "bdscript_decompiled") # just trust me lol
        input_fn_old = fn.replace("bdx_built", "bdscript_decompiled_old")
        original_fn = fn.replace("bdscript_decompiled", "bdx_built")
        outfn = os.path.join(outdir, str(random.randint(1000,90000))+f)
        outfn2 = os.path.join(outdir, str(random.randint(1000,90000))+f)
        outfnold = os.path.join(outdir, str(random.randint(1000,90000))+f)
        dirs = os.path.dirname(outfn)
        dirs2 = os.path.dirname(outfn2)
        dirsold = os.path.dirname(outfnold)
        if not os.path.isdir(dirs):
            os.makedirs(dirs)
        if not os.path.isdir(dirs2):
            os.makedirs(dirs2)
        if not os.path.isdir(dirsold):
            os.makedirs(dirsold)
        try:
            lib.openkh.bdx_compile(input_fn, outfn)
            if not os.path.exists(outfn):
                raise Exception("Compile failed!")
            lib.openkh.bdx_compile(input_fn, outfn2)
            if not os.path.exists(outfn2):
                raise Exception("Compile 2 failed")
            lib.openkh.bdx_compile(input_fn_old, outfnold)
            if not os.path.exists(outfnold):
                raise Exception("Old Compile failed")
        except:
            broken_builds.append(fn)
            
        if os.path.exists(outfn) and os.path.exists(outfnold):
            byte_identical[1] += 1
            original_bytes = bytearray(open(original_fn, "rb").read())
            new_bytes = bytearray(open(outfn, "rb").read())
            #DEBUG KINDA
            original_bytes = bytearray(open(outfnold, "rb").read())
            if original_bytes == new_bytes:
                byte_identical[0] += 1
```

```python
print("Num broken builds: {} / {}".format(len(broken_builds), c))
print("byte_identical: {} {}".format(byte_identical[0], byte_identical[1]))
```
