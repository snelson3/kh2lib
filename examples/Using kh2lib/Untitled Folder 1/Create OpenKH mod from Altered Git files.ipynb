```python
from kh2lib.kh2lib import kh2lib
import os, json, shutil
lib = kh2lib()
if os.path.isdir("mod"):
    shutil.rmtree("mod")
os.mkdir("mod")
```

```python
lib.gamedir = os.path.join(lib.gamedir, "KH2")
```

```python
TITLE="DJPRS Mod"
```

```python
lib.get_git_modifications()
```

```python
modyaml = {
    "assets": [
    ],
    "title": TITLE
}

for mod in lib.get_git_modifications():
    dest_path = os.path.join("files", mod)
    source_path = os.path.join(lib.gamedir, mod)
    if not os.path.exists(os.path.dirname(os.path.join("mod",dest_path))):
        os.makedirs(os.path.dirname(os.path.join("mod",dest_path)))
    shutil.copy(source_path, os.path.join("mod",dest_path))
    modyaml["assets"].append({
        "method": "copy",
        "name": mod,
        "source": [{"name":"files/"+mod}]
    })
import yaml
yaml.dump(modyaml, open(os.path.join("mod", "mod.yml"), "w"))
```
