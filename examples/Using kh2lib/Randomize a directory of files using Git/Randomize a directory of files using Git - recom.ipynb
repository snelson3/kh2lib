```python
lib = kh2lib(game="kh2")
lib.gamedir = 'C:\\Users\\12sam\\Desktop\\KH_Games\\Recom\\original'

outdir = os.environ["USE_OPENKH_MODS_DIR"]
# lib.reset_git()

# dir_to_randomize = os.path.join(os.environ["USE_KH2_GITPATH"], "KH2", "msn", "jp")
subdir = "CARD"
dir_to_randomize = os.path.join(lib.gamedir, subdir)

paths_to_randomize = [os.path.join(dir_to_randomize, i) for i in os.listdir(dir_to_randomize)]
paths_original_order = [f for f in paths_to_randomize]

import shutil, random

if os.path.exists(outdir):
    shutil.rmtree(outdir)
outdir = os.path.join(outdir, subdir)
os.makedirs(outdir)

random.shuffle(paths_to_randomize)
for i in range(len(paths_to_randomize)):
    old_path = paths_original_order[i]
    new_path = paths_to_randomize[i]
    old_basename = os.path.basename(old_path)
    new_basename = os.path.basename(new_path)
    output_path = os.path.join(outdir, old_basename)
    os.makedirs(output_path)
    if len(os.listdir(new_path)) > 1:
        continue
    for fn in os.listdir(new_path):
        old_name = os.listdir(old_path)[0] # only works when garuanteed 1 file
        shutil.copy(os.path.join(new_path, fn), os.path.join(output_path, old_name))
```
