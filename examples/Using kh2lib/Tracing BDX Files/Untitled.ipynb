```python
import os
modpath = "C:\\Users\\12sam\\Desktop\\openkh\\mods\\kh2\\thundrio-kh\\devmod\\files\\ai"
aifilename = "B_EX110_b_ex.bdscript"

trace_labels = True
trace_traps = True

infn = os.path.join(modpath, aifilename)
outfn = os.path.join(modpath, "new"+aifilename)

content = open(infn).read().split("\n")
new_content = []

found_txt = False
placed_strings = False
header_read = False

strings = {} # {labelname: string}

for l in range(len(content)):
    line = content[l]
    new_content.append(line)
    if not header_read:
        if "section .text" in line:
            header_read = True
    else:
        if not placed_strings and found_txt and line.strip()  == '':
            for k,v in strings.items():
                new_content.append('{}:'.format(k))
                new_content.append(" db '{}',0".format(v))
            placed_strings = True
        if not found_txt:
            if trace_labels and line.startswith("L") and not content[l+1].startswith("T"):
                labelname = line.split(":")[0]
                strings["LOG"+labelname] = labelname
                new_content.append(" pushFromPAi LOG{}".format(labelname))
                new_content.append(" syscall 0, 2 ; trap_puts")
                # Add the labels
            elif line.startswith("TXT"):
                found_txt = True
            elif trace_traps and "; trap" in line:
                trapname = "trap"+line.split("trap")[1].split(" ")[0] #I'm so lazy lol
                labelname = ''.join([c for c in trapname if c.isalpha()])
                strings[labelname] = trapname
                new_content.append(' pushFromPAi {}'.format(labelname))
                new_content.append(' syscall 0, 2 ; trap_puts;')
outcontent = '\n'.join(new_content)
open(outfn,"w").write(outcontent)
print(outcontent)
```
