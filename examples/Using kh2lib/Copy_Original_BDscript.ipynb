```python
import os,shutil

base_pth = 'C:\\Users\\12sam\\Desktop\\git\\kh2-ai-decompilation\\bdscript'
khbr_pth = 'C:\\Users\\12sam\\Desktop\\git\\khbr\\khbr\\KH2\\data\\bdscript'

for root, dirs, files in os.walk(khbr_pth):
    for ff in files:
        try:
            fn = os.path.join(root, ff)
            og_fn = fn.replace(khbr_pth,base_pth)
            out_fn = fn.replace(".bdscript", ".original.bdscript")
            shutil.copy(og_fn, out_fn)
        except:
            if "original" not in fn:
                print(fn)
```
