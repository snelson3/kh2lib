```python
class BinaryReader:
    def __init__(self, fn=None, data=None):
        if data:
            self.data = data
        else:
            self.data = bytearray(open(fn, "rb").read())
        self.pos = 0
    def readType(self, tpe, unpack=False):
        result = {
            "float32": self.readFloat32,
            "int32": self.readInt32,
            "int16": self.readInt16
        }[tpe]()
        if unpack and tpe == "float32":
            return self.unpackFloat32(result)
        return result
    def readByte(self):
        b = self.data[self.pos]
        self.pos += 1
        return b
    def readRest(self):
        arr = bytearray()
        for _ in range(len(self.data)-self.pos):
            arr.append(self.readByte())
        return arr
    def readFloat32(self):
        import struct
        bts = bytearray([self.readByte() for _ in range(4)])
        return bts
    def unpackFloat32(self, bts):
        return struct.unpack('<f', bts)[0]
    def readInt32(self):
        bts = self.data[self.pos:self.pos+4]
        self.pos += 4
        return int.from_bytes(bts, byteorder="little")
    def readInt16(self):
        bts = self.data[self.pos:self.pos+2]
        self.pos += 2
        return int.from_bytes(bts, byteorder="little")
```

```python
class BinaryWriter:
    def __init__(self, fn=None):
        self.fp = open(fn, "wb")
        self.nwrites = 0
    def writeType(self, tpe, data):
        return {
            "float32": self.writeFloat32,
            "int32": self.writeInt32,
            "int16": self.writeInt16
        }[tpe](data)
    def writeBytes(self, b):
        self.fp.write(b)
        self.nwrites += len(b)
        print([hex(bt) for bt in b])
    def writeArray(self, arr):
        for b in arr:
            self.writeBytes(bytes(b))
    def writeFloat32(self, value):
        packed_value = struct.pack('<f',value)
        self.writeBytes(packed_value)
    def writeInt32(self, value):
        self.writeBytes(int.to_bytes(value, length=4, byteorder="little"))
    def writeInt16(self, value):
        self.writeBytes(int.to_bytes(value, length=2, byteorer="little"))
    def close(self):
        self.fp.close()
```

```python
## Read PRTY
import os
fn = os.path.join(os.getcwd(), "workspace", "pref", "prty.list")
prty = BinaryReader(fn)
pointer_count = prty.readInt32()
pointers = [prty.readInt32() for _ in range(pointer_count)]
schema = [
    ("Walk Speed", "float32"),
    ("Run Speed", "float32"),
    ("Jump Height", "float32"),
    ("Turn Speed", "float32"),
    ("Hang Height", "float32"),
    ("Hang Margin", "float32"),
    ("Stun Time", "float32"),
    ("Mp Drive", "float32"),
    ("Up Down Speed", "float32"),
    ("Dash Speed", "float32"),
    ("Acceleration", "float32"),
    ("Brake", "float32"),
    ("Subjective Offset", "float32")
]
entries = []
for _ in range(33):
    ent = {}
    for ofs in schema:
        ent[ofs[0]] = prty.readType(ofs[1], unpack=True)
    entries.append(ent)
print(prty.pos)
footer = prty.readRest()
```

```python
for ent in entries:
    for k in ent:
        ent[k] = ent[k] * 3
```

```python
# Write PRTY
fn = os.path.join(os.getcwd(), "workspace", "pref", "prty.list")
new = BinaryWriter(fn)
new.writeInt32(pointer_count)
for i in range(pointer_count):
    new.writeInt32(pointers[i])
for ent in entries:
    for ofs in schema:
        new.writeType(tpe=ofs[1], data=ent[ofs[0]])
# new.writeBytes(footer)
new.close()
new.nwrites
```
