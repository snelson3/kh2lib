```python
class BinaryFile:
    def __init__(self, fn):
        self.data = bytearray(open(fn, "rb").read())
        self.pos = 0
    def readByte(self):
        b = self.data[self.pos]
        self.pos += 1
        return b
    def readInt32(self):
        bts = self.data[self.pos:self.pos+4]
        self.pos += 4
        return int.from_bytes(bts, byteorder="little")
    def readInt16(self):
        bts = self.data[self.pos:self.pos+2]
        self.pos += 2
        return int.from_bytes(bts, byteorder="little")
```
