```python
d1 = bytearray(open("KINGDOM HEARTS II FINAL MIX10.exe", "rb").read())
d2 = bytearray(open("KINGDOM HEARTS II FINAL MIX17.exe", "rb").read())
d1len = len(d1)
d2len = len(d2)
maxlen = max(d1len, d2len)
print("d2 is {} bytes larger than d1".format(d2len-d1len))
differences = 0
for i in range(maxlen):
    b1 = None if i >= d1len else d1[i]
    b2 = None if i >= d2len else d2[i]
    if b1 != b2:
        differences += 1
print("{} total differences".format(differences))
print("{}% different".format(int((differences/maxlen)*100)))
```
